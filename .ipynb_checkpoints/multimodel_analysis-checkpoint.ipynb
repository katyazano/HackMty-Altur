# 🛡️ Multi-Model Anti-Spoofing: AASIST + XGBoost Architecture
### Bank Telephony Voice Anti-Spoofing & Deepfake Detection
This interactive notebook explores and benchmarks a **Multi-Model Stacked Ensemble** combining:
1. **RawNet2**: SincNet + Residual GRU waveform baseline.
2. **AASIST-L**: SincNet + Spectro-Temporal Graph Attention Network (GAT).
3. **Acoustic XGBoost**: Gradient boosted decision trees on 136 handcrafted acoustic & spectral DSP features.
4. **AASIST + XGBoost Ensemble**: Stacked meta-learner combining 128-d graph embeddings with acoustic cues.

## 3. Audio Analysis & Interactive Audio Playback
Compare a genuine human bank call against a synthetic AI spoof sample.


```python
# Find one human sample and one synthetic sample from test set
test_dir = Path("Data/separated_agents/test/agent_0")
manifest_map = dict(zip(manifest_df["anon_id"], manifest_df["label"]))

human_sample = None
synth_sample = None

for f in sorted(list(test_dir.glob("*.wav"))):
    cid = f.stem.replace("_agent_0", "")
    lbl = manifest_map.get(cid)
    if lbl == "human" and human_sample is None:
        human_sample = f
    elif lbl == "synthetic" and synth_sample is None:
        synth_sample = f
    if human_sample and synth_sample:
        break

# Plot Waveforms & Spectrograms
fig, axes = plt.subplots(2, 2, figsize=(15, 8))

# Human
y_h, sr_h = librosa.load(human_sample, sr=16000)
librosa.display.waveshow(y_h, sr=sr_h, ax=axes[0, 0], color="#27ae60")
axes[0, 0].set_title(f"Authentic Human Voice ({human_sample.name})")

D_h = librosa.amplitude_to_db(np.abs(librosa.stft(y_h)), ref=np.max)
img_h = librosa.display.specshow(D_h, y_axis='hz', x_axis='time', sr=sr_h, ax=axes[0, 1], cmap='magma')
axes[0, 1].set_title("Human Spectrogram (Natural Harmonic Richness)")
fig.colorbar(img_h, ax=axes[0, 1], format="%+2.0f dB")

# Synthetic
y_s, sr_s = librosa.load(synth_sample, sr=16000)
librosa.display.waveshow(y_s, sr=sr_s, ax=axes[1, 0], color="#c0392b")
axes[1, 0].set_title(f"Synthetic AI Deepfake ({synth_sample.name})")

D_s = librosa.amplitude_to_db(np.abs(librosa.stft(y_s)), ref=np.max)
img_s = librosa.display.specshow(D_s, y_axis='hz', x_axis='time', sr=sr_s, ax=axes[1, 1], cmap='magma')
axes[1, 1].set_title("Synthetic Spectrogram (Vocoder & Phase Discontinuity)")
fig.colorbar(img_s, ax=axes[1, 1], format="%+2.0f dB")

plt.tight_layout()
plt.show()

print("🔊 Listen to Authentic Human Voice:")
display(Audio(data=y_h, rate=sr_h))

print("🔊 Listen to Synthetic AI Spoof Voice:")
display(Audio(data=y_s, rate=sr_s))
```

## 7. Interactive Live Prediction & Audio Tester
Test any bank call `.wav` recording on-the-fly and compare predictions across all models.


```python
# Load Trained Models into Memory
device = "cuda" if torch.cuda.is_available() else "cpu"

rawnet_m = RawNet2().to(device)
rawnet_m.load_state_dict(torch.load("weights/rawnet2_best.pth", map_location=device))
rawnet_m.eval()

aasist_m = AASIST().to(device)
aasist_m.load_state_dict(torch.load("weights/aasist_best.pth", map_location=device))
aasist_m.eval()

ensemble_m = AASISTXGBoostEnsemble(
    aasist_model=aasist_m,
    xgb_model_path="weights/xgboost_aasist_ensemble.json",
    scaler_path="weights/xgboost_scaler.joblib",
    device=device
)

def analyze_audio_call(audio_path: str):
    """Runs end-to-end multi-model analysis on a given audio file."""
    data, sr = sf.read(audio_path, dtype="float32")
    if data.ndim > 1:
        data = data[:, 0]
        
    t_in = torch.tensor(data, dtype=torch.float32).unsqueeze(0).to(device)
    
    # 1. RawNet2
    r_res = rawnet_m.predict_spoof(t_in)
    # 2. AASIST
    a_res = aasist_m.predict_spoof(t_in)
    # 3. Ensemble
    e_res = ensemble_m.predict_spoof(t_in, sample_rate=sr)
    
    print(f"=== Audio Analysis: {Path(audio_path).name} ===")
    display(Audio(data=data, rate=sr))
    
    res_table = pd.DataFrame([
        {"Model": "RawNet2", "Prediction": r_res["verdict"], "Human Confidence (%)": r_res["real_score_pct"], "Spoof Risk (%)": r_res["spoof_risk_pct"]},
        {"Model": "AASIST-L", "Prediction": a_res["verdict"], "Human Confidence (%)": a_res["real_score_pct"], "Spoof Risk (%)": a_res["spoof_risk_pct"]},
        {"Model": "AASIST + XGBoost Ensemble", "Prediction": e_res["verdict"], "Human Confidence (%)": e_res["real_score_pct"], "Spoof Risk (%)": e_res["spoof_risk_pct"]}
    ])
    display(res_table.style.map(lambda v: "color: red; font-weight: bold" if "SPOOF" in str(v) else ("color: green; font-weight: bold" if "HUMAN" in str(v) else "")))

# Test on human sample
analyze_audio_call(str(human_sample))

# Test on synthetic sample
analyze_audio_call(str(synth_sample))
```
